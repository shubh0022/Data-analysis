```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from statistics import mean, median, mode

# Load dataset
data = pd.read_csv('synthetic_student_marks.csv')

# Calculate mean, median, mode for 'Physics'
mean_val = data['Physics'].mean()
median_val = data['Physics'].median()
mode_val = data['Physics'].mode()[0]  # take first mode in case of multiple
print(mean_val)
print(median_val)
print(mode_val)

# Create histogram
fig = px.histogram(data_frame=data, x=['Physics'], nbins=20, title='Physics Marks: Histogram with Mean, Median, and Mode')

# Add vertical lines
fig.add_vline(x=mean_val, line_dash='dash', line_color='blue', annotation_text='Mean', annotation_position='top left')
fig.add_vline(x=median_val, line_dash='dot', line_color='green', annotation_text='Median', annotation_position='top right')
fig.add_vline(x=mode_val, line_dash='solid', line_color='red', annotation_text='Mode', annotation_position='bottom right')

# Show the plot
fig.update_layout(xaxis_title='Physics Marks', yaxis_title='Count')
fig.show()
```
